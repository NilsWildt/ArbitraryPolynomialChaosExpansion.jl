```python
import sympy as sy
```

```python
x, t, x2, t2, sigma, wt, wx, alpha = sy.var('x t x2 t2 σ2 ωt ωx α')
```

```python
k = sy.Function('k')(x, t, x2, t2, sigma, wt, wx)
```

```python
k = sigma*sy.exp(-0.5*(wx*(x-x2)**2+wt*(t-t2)**2))
```

```python
k
```

```python
dkdx = sy.Derivative(k,x,evaluate=True)
```

```python
dkdx2 = sy.Derivative(k,x2,evaluate=True)
```

```python
dkdt = sy.Derivative(k,t,evaluate=True)
```

```python
dkdt2 = sy.Derivative(k,t2,evaluate=True)
```

```python
d2kdxdx = sy.Derivative(dkdx,x,evaluate=True)
```

```python
d2kdx2dx2 = sy.Derivative(dkdx2,x2,evaluate=True)
```

```python
d2kdxdx2 = sy.Derivative(dkdx,x2,evaluate=True)
```

```python
k_uf =  sy.Derivative(k,t2,evaluate=True)-α*sy.Derivative(sy.Derivative(k,x2,evaluate=True),x2,evaluate=True)
```

```python
sy.print_rcode(k_uf.simplify())
```

```python
k_fu = sy.Derivative(k,t,evaluate=True)-α*sy.Derivative(sy.Derivative(k,x,evaluate=True),x,evaluate=True)
```

```python
sy.print_rcode(k_fu.simplify())
```
